```python
# Exercise 1 - Modulo examples

print("10 % 2 =", 10 % 2)   # remainder 0
print("10 % 3 =", 10 % 3)   # remainder 1
print("25 % 5 =", 25 % 5)   # remainder 0
print("25 % 6 =", 25 % 6)   # remainder 1
print("157 % 12 =", 157 % 12) # larger number example
print("99 % 4 =", 99 % 4)   # your choice
```

```python
# Exercise 2 - Modulo with user input

n = int(input("Enter a number (n): "))
m = int(input("Enter another number (m): "))

remainder = n % m
print(f"The remainder of {n} divided by {m} is {remainder}.")
```

```python
# Exercise 3 - Even or odd product

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

product = a * b
print("Product:", product)

if product % 2 == 0:
    print("The product is EVEN.")
else:
    print("The product is ODD.")
```

```python
# Exercise 4 - Fibonacci sequence

terms = int(input("How many Fibonacci terms? "))

# Start values
a, b = 1, 1

for i in range(terms):
    print(a)
    a, b = b, a + b
```
